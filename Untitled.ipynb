```python
""" Скрипт для построения 3d карты. 
Анализируется osa спектры двухволного лазера.

    """

import numpy as np
import matplotlib.pyplot as plt
import time
import datetime
import os
from tqdm import tqdm

from scripts.read_txt_file import read_txt_xy
from scripts.plot_dense_3d_surface import plot_dense_3d_surface
from scripts.plot_waterfall import plot_waterfall
from scripts.convert_dBm_mW import convert_dbm_to_mw
from scripts.plot_xy import plot_xy_markers
from scripts.write_data_txt import write_txt_xy
from scripts.plot_multiple_curves import plot_multiple_curves
from scripts.get_formated_time import get_formatted_time

VIEW_1 ={"elev": 30,          
        "azim": 90,         
        "roll": 0}
VIEW_2 ={"elev": 30,          
        "azim": 90,         
        "roll": 0}

LINEWIDTHS = [2]
CURRENT = 9.83
WAVELENGTH_START = 1056
WAVELENGTH_STOP = 1071  # Последний элемент не включается
WAVELENGTH_STEP = 1
WAVELENGTHS = np.arange(WAVELENGTH_START, WAVELENGTH_STOP, WAVELENGTH_STEP)

NUMBER_ITERATIONS=40
MEGA=1e+6
ITERATION_LABEL="Iteration (N)"
FREQ_LABEL_MHZ='Frequency (MHz)'
WAVE_LABEL='Wavelength (nm)'
POWER_LABEL_DBM = 'Power (dBm)'
POWER_LABEL_MW = "Power (mW)"



# PEAK = 'Yb'
PEAK = 'P205'


# Папка, где лежат данные 
DATA_FOLDER=rf'C:\Users\namys\Documents\DATA_dual_wavelength_laser\LD_set_current_9.83A_September-13-2026_time_10-18-08'

for linewidth in LINEWIDTHS:

        if PEAK == 'Yb':
            WAVE_START = 1000
            WAVE_STOP = 1100
            FWHM_WAVE=1.4

        if PEAK == 'P205':
            WAVE_START = 1200
            WAVE_STOP = 1300
            FWHM_WAVE=3.1
        
        SAVE_FOLDER=rf'{DATA_FOLDER}\OSA-3D-Graphs-{PEAK}\{}\{linewidth}nm'
        
    
        for wavelength in tqdm(WAVELENGTHS, desc=f"Processing LW {linewidth}nm"):
            
            
            power_data = []
            ampl_data = []
            wave_data = []
            
            for iteration in range(1, 41):
                osa_txt_file = rf"C:\Users\namys\Documents\DATA_dual_wavelength_laser\LD_set_current_9.83A_September-13-2026_time_10-18-08\OSA\linewidth_{linewidth}nm\wavelength_{wavelength}nm\osa_iteration_{iteration}_wavelength_{wavelength}nm_linewidth_{linewidth}nm_current_9.83A.txt"
                wave_arr, power_arr = read_txt_xy(path=osa_txt_file)
                
                
                # Перевод на линейный масштаб
                power_arr = convert_dbm_to_mw(power_arr_dbm=power_arr)

                # Перевод на микроВатт
                power_arr=power_arr*1e+6
                # Фильтруем по первичному диапазону (WAVE_START ... WAVE_STOP)
                mask_initial = (wave_arr >= WAVE_START) & (wave_arr <= WAVE_STOP)
                sub_wave_arr = wave_arr[mask_initial]
                sub_power_arr = power_arr[mask_initial]

                # Находим глобальный пик внутри этого диапазона
                max_index = np.argmax(sub_power_arr)
                max_wave = sub_wave_arr[max_index]

                # Строим новую маску вокруг max_wave, используя массив sub_wave_arr
                mask_peak = (sub_wave_arr >= max_wave - FWHM_WAVE) & (sub_wave_arr <= max_wave + FWHM_WAVE)

                # Фильтруем уже подрезанные массивы, чтобы получить окрестность пика
                final_wave_arr = sub_wave_arr[mask_peak]
                final_power_arr = sub_power_arr[mask_peak]
                max_pow = np.max(final_power_arr)
                

                # Добавляем в итоговый список
                power_data.append(final_power_arr)
                ampl_data.append(max_pow)
                wave_data.append(max_wave)
            # Генерация понятного заголовка для текущего графика
            avg_center_wave = np.average(wave_data)
            std_center_wave = np.std(wave_data)
            cv_center_wave = std_center_wave/avg_center_wave
            
            avg_max_ampl = np.average(ampl_data)
            std_max_ampl = np.std(ampl_data)
            cv_max_ampl = std_max_ampl/avg_max_ampl
            
            plot_title = (
                        f"LW filter: {linewidth}nm, WL filter: {wavelength}nm\n"
                        f'AVG WL: {avg_center_wave:.3f}, STD WL: {std_center_wave:.3f} CV: {cv_center_wave*100:.4f}%,\n'
                        f"AVG Ampl: {avg_max_ampl:.3f}, STD Ampl: {std_max_ampl:.3f}, CV: {cv_max_ampl*100:.4f}%")

            file_name = f'3D_plot_{wavelength}nm_{linewidth}nm_{PEAK}'
            plot_waterfall(
                x_arr=final_wave_arr,                
                y_arrays=power_data,               
                offsets=None,     
                x_label=WAVE_LABEL,
                y_label=None,
                z_label=POWER_LABEL_MW,
                title=None,
                elev=0,          
                azim=90,         
                roll=0,
                save_folder=SAVE_FOLDER,
                file_name=file_name,
                show_plot=False
            )
                
```
